```python
import sqlite3
import json
```

```python
def json_fetch(cur: sqlite3.Cursor):
    for line in cur:
        yield json.loads(line[0])
```

See [Sqlite's JSON query syntax](https://sqlite.org/draft/json1.html#jexb)

```python
conn = sqlite3.connect("../wikipedia.db")
cur = conn.cursor()
def json_query(query:str, limit=2):
    return (json.loads(line[0]) for line in cur.execute(f"SELECT data -> ? FROM movie LIMIT ?;", (query,limit)))
```

```python
for line in json_query("$.synopsis"):
    print(line)
    print()
```
